# 🌊 Staggered Letter Wave Hover Animation
This notebook documents how to create a high-end, interactive staggered letter wave hover effect using **React**, **Tailwind CSS**, and **Framer Motion**.

## 🎯 Objective
When a user hovers over a button or navigation link, each letter should sequentially slide up, fade out, and be replaced by an orange version sliding in from below, creating a smooth wave effect.

## 💡 Step 2: The Core Concept

The wave animation relies on three main techniques:
1. **String Splitting**: We break the target text string into an array of individual characters (`text.split("")`).
2. **Dual-Layer Positioning**: Two text layers rendered on top of each other inside an `overflow-hidden` container.
   - **Layer 1 (Default)**: Slides up from `y: 0` to `y: -20` on hover.
   - **Layer 2 (Orange)**: Positioned with `absolute inset-0`, slides up from `y: 20` to `y: 0` on hover.
3. **Staggered Delays**: Each character index `i` receives a delay calculation `delay: i * 0.03` to stagger the motion.

## 🧩 Step 3: Reusable Component Code

Here is the standalone TypeScript/React component implementation:

```javascript
import React from 'react';
import { motion } from 'framer-motion';

interface WaveTextProps {
  text: string;
  activeColor?: string;
  defaultColor?: string;
}

export const WaveText: React.FC<WaveTextProps> = ({
  text,
  activeColor = "#FF3917",
  defaultColor = "#FFFFFF"
}) => {
  const letters = text.split("");

  return (
    <div className="relative overflow-hidden block select-none">
      {/* Default Text Layer (Slides Up & Out) */}
      <div className="flex whitespace-nowrap">
        {letters.map((char, i) => (
          <motion.span
            key={i}
            variants={{
              initial: { y: 0, opacity: 1, color: defaultColor },
              hover: { y: -20, opacity: 0, color: activeColor }
            }}
            transition={{
              duration: 0.2,
              delay: i * 0.03,
              ease: "easeInOut"
            }}
            className="inline-block"
          >
            {char === " " ? "\u00A0" : char}
          </motion.span>
        ))}
      </div>

      {/* Active Replacement Layer (Slides Up & In) */}
      <div className="absolute inset-0 flex whitespace-nowrap">
        {letters.map((char, i) => (
          <motion.span
            key={i}
            variants={{
              initial: { y: 20, opacity: 0, color: activeColor },
              hover: { y: 0, opacity: 1, color: activeColor }
            }}
            transition={{
              duration: 0.2,
              delay: i * 0.03,
              ease: "easeInOut"
            }}
            className="inline-block"
          >
            {char === " " ? "\u00A0" : char}
          </motion.span>
        ))}
      </div>
    </div>
  );
};
```

## 🚀 Step 4: Integration Example

Wrap `WaveText` inside a Framer Motion button using `initial="initial"` and `whileHover="hover"` to propagate hover states down to child spans.

```javascript
export const WaveButton = () => {
  return (
    <motion.button
      initial="initial"
      whileHover="hover"
      className="bg-black text-white font-mono text-xs uppercase tracking-widest px-5 py-3 cursor-pointer inline-flex items-center gap-2 overflow-hidden"
    >
      <WaveText text="ABOUT US" defaultColor="#FFFFFF" activeColor="#FF3917" />
      
      {/* Arrow Icon with diagonal hover translation */}
      <motion.span
        variants={{
          initial: { x: 0, y: 0, color: "#FFFFFF" },
          hover: { x: 2, y: -2, color: "#FF3917" }
        }}
        transition={{ duration: 0.2, ease: "easeInOut" }}
        className="text-sm leading-none inline-block"
      >
        ↗
      </motion.span>
    </motion.button>
  );
};
```

## 🔑 Key Animation Attributes Breakdown

| Class / Property | Purpose |
|---|---|
| `text.split("")` | Converts the string into an iterable array of characters. |
| `delay: i * 0.03` | Creates the sequential wave motion across individual letters. |
| `char === " " ? "\u00A0" : char` | Replaces standard spaces with non-breaking spaces so layout width is maintained. |
| `overflow-hidden` | Clips text moving outside the boundary of the link or button container. |
| `whileHover="hover"` | Parent directive that triggers the `hover` variant on all nested `motion` children. |
